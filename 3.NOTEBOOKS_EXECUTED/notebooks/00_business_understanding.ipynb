## I. Giới thiệu dự án
Ngành công nghiệp âm nhạc toàn cầu đang trải qua sự bùng nổ dữ liệu vô tiền khoáng hậu. Mỗi ngày, hàng chục ngàn bài hát mới được phát hành trên các nền tảng streaming như Spotify, Apple Music. Giữa một đại dương âm thanh đó, việc nhận diện, đánh giá và dự báo một ca khúc có tiềm năng trở thành "Hit" (bài hát ăn khách) không còn chỉ dựa vào cảm nhận nghệ thuật thuần túy, mà đòi hỏi sức mạnh của Khoa học Dữ liệu (Data Science).

**HitRadar Pro** là một hệ thống AI hỗ trợ ra quyết định dành cho các nhà sản xuất âm nhạc (Producers), hãng thu âm (Record Labels) và các nghệ sĩ độc lập. Hệ thống sử dụng Machine Learning để phân tích các đặc trưng âm thanh và metadata, từ đó dự báo mức độ phổ biến của một bản thu âm.


## II. Tìm hiểu nghiệp vụ Streaming Âm nhạc
Trong hệ sinh thái Spotify, sự phổ biến của một bài hát bị chi phối bởi thuật toán đề xuất (Recommendation Algorithm). Thuật toán này không chỉ nhìn vào tên tuổi nghệ sĩ mà còn "lắng nghe" sâu vào các đặc tính toán học của file audio:
- **Động lực học (Dynamics):** Năng lượng (Energy), Độ lớn (Loudness), Tính có thể nhảy (Danceability).
- **Cấu trúc (Structure):** Nhịp độ (Tempo), Cung điệu (Key), Loại nhạc (Acousticness, Instrumentalness).
Một bài hát ăn khách hiện đại thường phải tuân thủ những "luật chơi" ngầm của thuật toán (ví dụ: thời lượng ngắn lại để tăng số lượt stream, năng lượng cao ở 15 giây đầu tiên).


## III. Bài toán HitRadar và biến TARGET
**Bài toán:** Xây dựng mô hình Machine Learning Hồi quy (Regression) hoặc Phân lớp (Classification) để dự đoán mức độ thành công của một bài hát trước khi hoặc ngay khi nó được phát hành.

**Biến mục tiêu (TARGET): `popularity`**
- Đây là một thang điểm từ **0 đến 100** do chính Spotify cung cấp, phản ánh mức độ phổ biến của bài hát ở thời điểm hiện tại.
- Thuật toán của Spotify tính toán điểm này chủ yếu dựa trên tổng số lượt phát (plays) và mức độ gần đây của những lượt phát đó (bài hát đang hot hiện tại sẽ có điểm cao hơn bài hát hot trong quá khứ).


## IV. Mục tiêu nghiên cứu
1. **Khám phá (Explore):** Phân tích sự thay đổi trong gu âm nhạc của công chúng từ năm 1920 đến 2021.
2. **Giải mã (Decode):** Tìm ra mối tương quan thực sự giữa các đặc trưng âm thanh (như độ ồn, nhịp độ) với độ phổ biến của bài hát.
3. **Dự báo (Predict):** Huấn luyện mô hình AI (như Random Forest, XGBoost) có khả năng dự đoán điểm `popularity` với độ chính xác cao.
4. **Ứng dụng (Deploy):** Đưa mô hình vào môi trường thực tế (Gradio/FastAPI) để người dùng có thể nhập thông số âm thanh và nhận về dự báo ngay lập tức.


## V. Lý do chọn dataset và phân tích SWOT
**Lý do chọn Spotify Dataset:**
Bộ dữ liệu gồm hơn 580,000 bài hát với 20 thuộc tính (thu thập qua Spotify API) là một trong những tập dữ liệu đồ sộ và chuẩn mực nhất về âm thanh học (Acoustic properties) hiện có trên Kaggle. Nó chứa đầy đủ các khía cạnh: Metadata (Tên, Năm, Nghệ sĩ) và Audio Features (Danceability, Energy...).

**Phân tích SWOT của dự án HitRadar:**
- **S (Strengths):** Tập dữ liệu khổng lồ, feature đa dạng. Ứng dụng thực tế cao cho các nhà sản xuất âm nhạc.
- **W (Weaknesses):** Điểm Popularity của Spotify có tính thời điểm (thiên vị những bài hát mới phát hành), làm khó khăn trong việc dự đoán khách quan giá trị cốt lõi của bài hát.
- **O (Opportunities):** Có thể mở rộng tích hợp thành Plugin cho các phần mềm làm nhạc (DAW như FL Studio, Ableton).
- **T (Threats):** Gu âm nhạc của công chúng liên tục thay đổi (Data Drift), mô hình cần được huấn luyện lại thường xuyên với dữ liệu mới.


## VI. Dataset và công nghệ sử dụng
- **Nguồn dữ liệu:** Dữ liệu Raw dạng CSV tải về từ Kaggle/Spotify API.
- **Lưu trữ & Pipeline:** PostgreSQL (Raw Data $\rightarrow$ Clean Data $\rightarrow$ Analytics Views).
- **Ngôn ngữ:** Python 3.x
- **Khám phá & Tiền xử lý (EDA & Preprocessing):** Pandas, Numpy, Matplotlib, Seaborn, Scikit-learn.
- **Mô hình hóa (Modeling):** XGBoost, Random Forest, LightGBM (Ensemble Methods).
- **Triển khai (Deployment):** Joblib/Pickle (Lưu mô hình), Gradio (Giao diện UI), FastAPI (Backend API).


## VII. Tình huống thực tế, vai trò AI và giá trị hệ thống
**Tình huống (Use Case):**
Một nghệ sĩ Indie vừa hoàn thành xong bản Demo của ca khúc mới. Họ phân vân không biết bài hát này có hợp "trend" thị trường không, hay nó quá "kén" người nghe.

**Giải pháp với HitRadar:**
Nghệ sĩ trích xuất các đặc trưng âm thanh của bản Demo, nhập vào hệ thống HitRadar. AI phân tích và đưa ra kết quả: *"Bài hát này có độ Popularity dự kiến là 65/100. Lời khuyên: Hãy tăng Tempo (Nhịp độ) lên thêm 10 BPM và làm phần Bass Loudness mạnh hơn một chút để tiếp cận tệp khán giả hiện đại tốt hơn."*

**Giá trị (Value):**
Tiết kiệm chi phí Marketing khổng lồ bằng cách tối ưu hóa "chất liệu lõi" (bản thân bài hát) trước khi chi tiền chạy quảng cáo.


## VIII. Tổng kết
Dự án **HitRadar Pro** không chỉ là một bài toán phân tích dữ liệu đơn thuần mà là một giải pháp AI toàn diện đóng vai trò như một "nhà cố vấn âm nhạc số". Bằng việc bám sát vào quy trình Data Science chuẩn mực (Từ việc hiểu nghiệp vụ, khám phá dữ liệu, làm sạch, kỹ thuật đặc trưng cho đến triển khai mô hình), hệ thống hứa hẹn sẽ mang lại góc nhìn khoa học định lượng (Quantitative) cho một lĩnh vực vốn được coi là hoàn toàn định tính (Qualitative) như âm nhạc.
